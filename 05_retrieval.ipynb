```python
evaluation_questions = [
    "What are the main objectives of the project?",
    "What is the motivation behind the project?",
    "What technologies are integrated into the system?",
    "What is the target end-to-end latency?",
    "What datasets were used for evaluation?"
]

print("Number of questions:", len(evaluation_questions))
```

```python
for question in evaluation_questions:

    query_embedding = generate_embeddings([question])

    results = retrieve(
        index,
        query_embedding,
        chunks,
        k=3
    )

    print("\nQUESTION:", question)

    for result in results:
        print(
            "Page:", result["page"],
            "| Distance:", round(result["distance"], 3)
        )

    print("-" * 80)
```

```python
evaluation_results = [
    {
        "question": "What are the main objectives of the project?",
        "relevant": True
    },
    {
        "question": "What is the motivation behind the project?",
        "relevant": True
    },
    {
        "question": "What technologies are integrated into the system?",
        "relevant": True
    },
    {
        "question": "What is the target end-to-end latency?",
        "relevant": True
    },
    {
        "question": "What datasets were used for evaluation?",
        "relevant": True
    }
]

relevant_count = sum(
    result["relevant"]
    for result in evaluation_results
)

accuracy = relevant_count / len(evaluation_results)

print("Retrieval Accuracy:", accuracy)
```
